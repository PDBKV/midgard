# IGS metadata parser
The IGS satellite metadata file as defined in [this pdf](https://files.igs.org/pub/resource/working_groups/multi_gnss/Metadata_SINEX_1.10.pdf) it provides information about different identifiers, PRN, frequency and plane allocations, satellite masses and geometries.

The parser is imported and the example file is parsed.

```python
from midgard import parsers

p = parsers.parse_file(parser_name="igs_metadata", file_path="../examples/parsers/igs_metadata.snx")

for key, value in p.data.items():
    print(key, ":\n")
    for key, value in value.items():
        print(key, ":\n")
        for key, value in value.items():
            print(key, " : ", value)
```
